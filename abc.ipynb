```python
#question 3(a)
n = int(input("Enter number of elements: "))

a = []

for i in range(n):
    x = float(input("Enter numbers: "))
    a.append(x)

mean = sum(a)/n

variance = 0

for i in a:
    variance = variance + (i-mean)**2

variance = variance/n

sd = variance**0.5

print("Mean =", mean)
print("Variance =", variance)
print("Standard Deviation =", sd)
```

```python
#question 4
f = open("input.txt", "r")

data = f.read()

f.close()

words = data.split()

d = {}

for i in words:
    if i in d:
        d[i] = d[i] + 1
    else:
        d[i] = 1

a = list(d.items())

a.sort(key=lambda x: x[1], reverse=True)

print("10 Most Frequent Words:")

for i in a[:10]:
    print(i[0], ":", i[1])
```

```python
#question 5
a = []

for i in range(6):
    x = int(input("Enter marks: "))
    a.append(x)
print("Original Marks:", a)
for i in range(6):
    for j in range(5-i):
        if a[j] < a[j+1]:
            x = a[j]
            a[j] = a[j+1]
            a[j+1] = x

print("Marks in descending order:")

for i in a:
    print(i)
```

```python
#question 6
f = open("input1.txt", "r")

a = f.readlines()

f.close()

b = []

for i in a:
    x = i.strip()
    b.append(x)

b.sort()

f = open("output.txt", "w")

for i in b:
    f.write(i + "\n")

f.close()

print("Sorted contents written to output.txt")
```

```python
#question 7
def DivExp(a, b):
    assert a > 0, "a must be positive"

    if b == 0:
        raise ZeroDivisionError("Division by zero is not allowed")

    c = a / b
    return c

a = int(input("Enter a: "))
b = int(input("Enter b: "))

try:
    c = DivExp(a, b)
    print("Result =", c)

except Exception as e:
    print("Error:", e)

```

```python
#question 8 
class Complex:
    def __init__(self, r, i):
        self.r = r
        self.i = i

    def add(self, c):
        return Complex(self.r + c.r, self.i + c.i)

    def display(self):
        print(self.r, "+", self.i, "i")


n = int(input("Enter number of complex numbers: "))

c = Complex(0, 0)

for i in range(n):
    r = int(input("Enter real part: "))
    im = int(input("Enter imaginary part: "))

    x = Complex(r, im)
    print("Complex number:", r, "+", im, "i")
    c = c.add(x)

print("Sum of complex numbers:")
c.display()
```

```python
#question 9 
n = int(input("Enter number of students: "))

students = []

for i in range(n):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))

    students.append((name, marks))

total = 0

for i in students:
    total = total + i[1]

average = total / n

topper = max(students, key=lambda x: x[1])

print("\nStudent Details:")

for i in students:
    print(i[0], ":", i[1])

print("Average Marks =", average)
print("Topper =", topper[0])
print("Highest Marks =", topper[1])
```

```python
#Question 10 
import os

def display(path):

    for i in os.listdir(path):

        fullpath = os.path.join(path, i)

        if os.path.isdir(fullpath):
            print(i, "- Directory")
            display(fullpath)

        else:
            print(i, "- File")


path = input("Enter folder path: ")

display(path)
```
